# Phân tích phân phối và quan hệ dữ liệu

Notebook này thực hiện Câu 3, Câu 4, Câu 8 và Câu 2.c trên bảng dữ liệu đã cho.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

data = pd.DataFrame({
    "ID": range(1, 11),
    "Tuổi": [25, 30, 22, 40, 35, None, 29, 50, 27, 31],
    "Lương": [8000, None, 5000, 15000, -2000, 7000, 8500, 30000, 8200, 7800],
    "Điểm": [7.5, 8.0, 6.5, 9.0, 7.0, 6.0, 8.5, 9.5, None, 7.8],
    "Giới tính": ["Nam", "Nữ", "Nam", "Nữ", "Nam", "Nữ", "Nam", "Nữ", "Nam", "Nữ"],
    "Mua hàng": ["Yes", "No", "Yes", "Yes", "No", "Yes", "Yes", "Yes", "No", "Yes"]
})

data
```

## Câu 3: Phân tích phân phối

### Histogram của Tuổi và Boxplot của Lương

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(data=data, x="Tuổi", bins=7, kde=True, ax=axes[0], color="#2878b5")
axes[0].set_title("Histogram phân phối Tuổi")
axes[0].set_xlabel("Tuổi")
axes[0].set_ylabel("Tần số")

sns.boxplot(y=data["Lương"], ax=axes[1], color="#e67e22")
axes[1].set_title("Boxplot phân phối Lương")
axes[1].set_ylabel("Lương")

plt.tight_layout()
plt.show()
```

## Câu 4: Phân tích quan hệ

### Scatter Plot giữa Tuổi và Lương

Chỉ các dòng có đồng thời Tuổi và Lương được dùng để vẽ biểu đồ.

```python
scatter_data = data.dropna(subset=["Tuổi", "Lương"])

plt.figure(figsize=(8, 5))
sns.scatterplot(data=scatter_data, x="Tuổi", y="Lương", s=100, color="#16a085")
plt.title("Scatter Plot giữa Tuổi và Lương")
plt.xlabel("Tuổi")
plt.ylabel("Lương")
plt.tight_layout()
plt.show()

print(f"Hệ số tương quan Pearson: {scatter_data['Tuổi'].corr(scatter_data['Lương']):.3f}")
```

## Câu 8 và Câu 2.c: Phát hiện outlier bằng IQR

Quy tắc IQR:

- $IQR = Q3 - Q1$
- Cận dưới $= Q1 - 1.5 \times IQR$
- Cận trên $= Q3 + 1.5 \times IQR$
- Giá trị nằm ngoài hai cận là outlier.

```python
def iqr_report(series, name):
    values = series.dropna()
    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = values[(values < lower_bound) | (values > upper_bound)]

    print(f"{name}")
    print(f"Q1 = {q1:.2f}, Q3 = {q3:.2f}, IQR = {iqr:.2f}")
    print(f"Cận dưới = {lower_bound:.2f}, Cận trên = {upper_bound:.2f}")
    print(f"Outlier: {outliers.tolist() if not outliers.empty else 'Không có'}")
    return outliers

salary_outliers = iqr_report(data["Lương"], "Câu 8 - Lương")
age_outliers = iqr_report(data["Tuổi"], "Câu 2.c - Tuổi")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.boxplot(y=data["Lương"], ax=axes[0], color="#e67e22")
axes[0].set_title("Câu 8 - Boxplot Lương")
axes[0].set_ylabel("Lương")

sns.boxplot(y=data["Tuổi"], ax=axes[1], color="#2878b5")
axes[1].set_title("Câu 2.c - Boxplot Tuổi")
axes[1].set_ylabel("Tuổi")

plt.tight_layout()
plt.show()
```

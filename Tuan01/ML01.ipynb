### Bài 9. Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```

### Bài 10. Giải hệ phương trình bằng Autograd

```python
import torch

A = torch.tensor(1.0, requires_grad=True)
B = torch.tensor(1.0, requires_grad=True)
C = torch.tensor(1.0, requires_grad=True)
D = torch.tensor(1.0, requires_grad=True)

optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for epoch in range(3000):
    y1 = A + B - 9       # A + B = 9
    y2 = C - D - 1       # C - D = 1
    y3 = A + C - 8       # A + C = 8
    y4 = B - D - 2       # B - D = 2

    loss = y1**2 + y2**2 + y3**2 + y4**2

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

print("A =", A.detach().item())
print("B =", B.detach().item())
print("C =", C.detach().item())
print("D =", D.detach().item())
print("Sai số =", loss.detach().item())
```

### Bài 11. Viết chương trình PyTorch để thực hiện các yêu cầu sau:

```python
import torch

# 1. Reshape Tensor X
X = torch.arange(24)

print("X ban đầu:", X)
print("X dạng 4 x 6:\n", X.reshape(4, 6))
print("X dạng 2 x 12:\n", X.reshape(2, 12))
print("X dạng 2 x 3 x 4:\n", X.reshape(2, 3, 4))

# 2. Các phép toán trên A và B
A = torch.rand(3, 4) + 0.5
B = torch.rand(3, 4) + 0.5

print("\nA:\n", A)
print("\nB:\n", B)
print("\nA + B:\n", A + B)
print("\nA * B:\n", A * B)
print("\nA / B:\n", A / B)
print("\nTrung bình theo cột:\n", A.mean(dim=0))
print("\nĐộ lệch chuẩn theo cột:\n", A.std(dim=0))

# 3. Autograd với y = 2x^2 + 5x + 3
x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
y.backward()

dy_dx_autograd = x.grad
dy_dx_manual = 4 * 2 + 5

print("\ny =", y.item())
print("dy/dx bằng autograd =", dy_dx_autograd.item())
print("dy/dx thủ công =", dy_dx_manual)
```

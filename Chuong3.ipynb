# Creating Tensors

```python
import torch
a = torch.tensor([[1,2,3], [4,5,6]], dtype=torch.int32)
print(a)
```

```python
import torch
a = torch.linspace(-1, 1, 10)
print(a)
```

```python
import torch
a = torch.rand(3,4)
print(a)
```

```python
import torch
a = torch.randint(3, 10, size=(3,4))
print(a)
```

```python
import torch
a = torch.zeros(2, 3, 4)
print(a)
```

```python
import torch
a = torch.full((2,3,4), 5)
print(a)
```

```python
import torch
a = torch.eye(4)
print(a)
```

```python
import torch
a = torch.zeros(2, 3, 4)
print(a.shape)
print(a.size())
```

```python
print(a.ndim)
```

```python
import torch
a = torch.zeros(2, 3, 4)
print(len(a))
```

```python
print(a.dtype)
```

```python
import torch
a = torch.zeros(2, 3, 4)
b = a.type(torch.int32)
print(a.dtype)
print(b.dtype)
```

```python
a = torch.randn(3,4,5)
print(a)
```

```python
print(a[1])
```

```python
print(a[1:, 2:4])
```

```python
print(a[:, None, :, None].shape)
```

```python
b = torch.unsqueeze(a, dim=2)
print(a.shape)
print(b.shape)
```

```python
import torch
a = torch.randn(3,4)
print(a)
print(a[:, (a > -1).all(axis=0)])
```

```python
print(a[:, [1,0,0,1]])
```

```python
import torch
a = torch.randn(3,4)
print(a)
print(a.ravel())
```

```python
print(a.reshape(-1))
```

```python
print(a.reshape(3,2,2))
```

```python
print(a.T)
```

```python
print(a.transpose(0, 1))
```
